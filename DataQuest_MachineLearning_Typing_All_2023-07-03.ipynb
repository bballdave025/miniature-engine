### DQ 01.02 Data Collection and Exploration

```python
### DQ 01.02 Data Collection and Exploration

import pandas as pd
from sklearn.datasets import load_breast_cancer

cancer_data = load_breast_cancer(as_frame=True)
cancer_df = cancer_data.data
cancer_df['target'] = cancer_data.target

print()
print("First few observations:")
print(cancer_df.head())

print()
print("Shape:")
print(cancer_df.shape)

n_stated_observations = 569
assert(cancer_df.shape[0] == n_stated_observations)

n_header_rows = 1
n_stated_features = 30
assert(cancer_df.shape[1] == n_header_rows + n_stated_features)

print()
print("If we're here without any AssertionError stuff,")
print("we have confirmed that the number of features")
print("and observations are [the] same as the ones")
print("listed above.")

do_show_by_column_dwb = True

if do_show_by_column_dwb:
    print()
    for col_name in cancer_df.columns.values:
        print( (f"col_name: {col_name}; "
                f"n_nan: {cancer_df[col_name].isna().sum()}"
               )
             )
    ##endof:  for col_name
##endof:  if do_show_by_column_dwb

print()
print("Total NAN: " + str(cancer_df.isna().sum().sum()))

print()
print("Summary NAN: ")
print(cancer_df.isna().sum())

print("Number with each classification:")
print(cancer_df["target"].value_counts())

print()
print("Summary_statistics:")
print(cancer_df)
```
